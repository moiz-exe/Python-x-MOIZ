***STRING CONCATENATON AND REPETITION***

```python
full_name = 'Usman' + ' Ghani' #concatenation
print(full_name)
full_name = 'Usman' + ' ' + 'Ghani' #another syntax
print(full_name)

name = 'Usman ' * 3 #repetition
print(name)         #prints 3 times 'Usman'

```

Concatenation refers to the operation of joining two or more strings end-to-end to form a single new string. In Python, you can achieve this using the + operator. As demonstrated in the notebook above. I assigned a variable of full_name to 'Usman' + ' Ghani' and by printing it displays Usman Ghani.
And if change the syntax . As i wrote 'Usman' +''+ 'Ghani'.

**Slicing String**

```python
str = "program"    #case1
r = str[0:2]
print(r)

r2 = str[2:]       #case2
print(r2)

r3 = str[2:5]      #case3
print(r3)
```

**Case1**
String also supports slicing.
which means getting characters in a specified range
for example just take a look on above program.
i initialized a string in a variable str and than i used **slicing**
by defining the range of the variable containing string to a new variable.
Note: Range includes first index but excludes last one.
AS, i defined the range str[0:2] it doesn't includes the character on index 2.
**Case2**
if i do not specify the ending point of range e.g if i write r2 = str[2:] it starts from index 2 and continues till the end.

**Case 3**
If i define the range as [2:5] as mentioned in the program
it prints from index 2 to 4 it do not prints the ending index point.


It is how we use slicing. I hope you understand.
